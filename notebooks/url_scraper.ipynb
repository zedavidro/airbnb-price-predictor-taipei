```python
import requests
import pandas as pd
import numpy as np
from PIL import Image
from io import BytesIO
import os
import random
import matplotlib.pyplot as plt
import time
```

```python
dir = 'data/images'
df = pd.read_csv("data/listings.csv")
url_list = df['picture_url']
```

```python
for index, url in enumerate(url_list):
    save_path = f"{dir}/{index}.jpg"
    download_and_resize_image(url, (256,256), save_path)
    time.sleep(0.5)
```

```python
def download_and_resize_image(url, size=(256, 256), save_path='image.jpg'):
    try:
        response = requests.get(url)
        response.raise_for_status()
        
        img = Image.open(BytesIO(response.content))
        img = img.convert('RGB')
        img = img.resize(size, Image.Resampling.LANCZOS)

        img.save(save_path)
        print(f"Image saved to {save_path}")

    except Exception as e:
        print(f"Error: {e}")
```

#### Pictures

```python
image_dir = "data/images"
sample_size = 220
cols = 22
rows = (sample_size + cols - 1) // cols

image_paths = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith((".jpg", ".png"))]
sample_paths = random.sample(image_paths, sample_size)

plt.figure(figsize=(cols * 2, rows * 2))
for i, path in enumerate(sample_paths):
    img = Image.open(path)
    plt.subplot(rows, cols, i + 1)
    plt.imshow(img)
    plt.axis('off')
plt.subplots_adjust(wspace=0.05, hspace=0.05)
plt.show()
```
