```python
import pandas as pd
```

```python
df = pd.read_csv("/kaggle/input/listings/listings.csv")
df.head()
```

```python
df_trimmed = df[["id", "description"]]
df_trimmed.head(15)
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')
```

```python
descriptions = df_trimmed["description"].dropna().tolist()
embeddings = model.encode(descriptions, show_progress_bar=True)
```

```python
df_trimmed = df_trimmed.loc[df_trimmed["description"].notnull()]
df_trimmed["embedding"] = embeddings.tolist()
```
